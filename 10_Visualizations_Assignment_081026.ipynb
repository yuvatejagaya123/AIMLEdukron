```python
# 1. Matplotlib

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, color='blue', marker='o', linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
# plt.plot() - line charts
# plt.bar() - bar charts
# plt.scatter() - scatter plots
# plt.hist() - histograms
# plt.boxplot() - box plots
# plt.pie() - pie charts
```

## 📊 Short Insights

- The graph shows a **steady upward trend** from 10 to 40.
- Values increase consistently as the X-axis values increase.
- The highest value is **40** at X = 5.
- The lowest value is **10** at X = 1.
- No declines or fluctuations are observed throughout the trend.
- Overall, the data indicates **continuous and stable growth**.

```python
## Simple customization

plt.figure(figsize=(8, 5))
plt.bar(x, y, linestyle='--', color='red', label='Sales')
plt.legend()
plt.tight_layout()
plt.show()
```

## 📊 Short Insights

- The bar chart shows a **steady increase in sales** from 10 to 40.
- Sales grow consistently across all categories/time periods.
- The highest sales value is **40** at category **5**.
- The lowest sales value is **10** at category **1**.
- The largest increase is observed between **1 → 2** and **4 → 5**.
- Overall, the chart indicates **strong and continuous sales growth**.

```python
# 2. Seaborn

import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
flights = sns.load_dataset('flights')

sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```

```python
# sns.lineplot() - line chart
# sns.barplot() - bar chart
# sns.scatterplot() - scatter plot
# sns.histplot() - histogram
# sns.boxplot() - box plot
# sns.heatmap() - heat map
# sns.countplot() - count plots
```

## 📊 Short Insights

- The graph shows a strong upward trend in passenger numbers over time.
- Passenger count increases consistently from around **130** in 1949 to nearly **480** in 1960.
- The highest passenger volume is recorded in **1960**.
- Growth accelerates after **1955**, indicating increasing demand.
- No major declines are observed throughout the period.
- Overall, the data suggests **steady and continuous growth in air travel passengers**.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Bill Amount by Day')
plt.show()
```

## 📊 Short Insights

- **Sunday** has the highest median bill amount, followed closely by **Saturday**.
- **Saturday and Sunday** show greater variability in bill amounts compared to weekdays.
- **Thursday and Friday** have relatively lower bill amounts and a smaller spread.
- Several **outliers** are present, especially on **Saturday and Sunday**, indicating some unusually high bills.
- The highest bill amounts are observed on **Saturday** (above 50).
- Overall, customers tend to spend **more on weekends than on weekdays**.

```python
# 3. Plotly

import plotly.express as px

df = px.data.iris()

fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

## 📊 Short Insights

- The scatter plot compares **sepal width** and **sepal length** for three Iris flower species.
- **Virginica** flowers generally have the largest sepal lengths, while **Setosa** has the smallest.
- **Versicolor** falls between Setosa and Virginica in terms of sepal size.
- Setosa points form a relatively distinct cluster, whereas Versicolor and Virginica show some overlap.
- Most observations are concentrated between **2.5-3.5 sepal width** and **5.0-7.5 sepal length**.
- The plot shows that species can be differentiated based on their sepal measurements, although some overlap exists.
