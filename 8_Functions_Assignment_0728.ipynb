```python
# Assignment 17: Palindrome Checker
# Write a function named is_palindrome().

# Requirements
# Accept a string.
# Ignore spaces and letter case.
# Return True when the string is a palindrome.
# Print a meaningful result.

def is_palindrome(text):
  text = text.lower().replace(" ","")
  return text == text[::-1]

text = str(input("Enter text: "))
if is_palindrome(text):
  print("The text is a palindrome.")
else:
  print("The text is not a palindrome.")
```

```python
# Assignment 20: Count Positive, Negative and Zero Values
# Write a function named count_number_types().

# Requirements
# Accept a list of integers.
# Count positive numbers, negative numbers, and zeros.
# Return all three counts.
# Use tuple unpacking to receive the returned values.

def count_number_types(numbers):
  positive = 0
  negative = 0
  zero = 0

  for num in numbers:
    if num > 0:
      positive += 1

    elif num == 0:
      zero += 1

    elif num < 0:
      negative += 1

  return positive, negative, zero

numbers = list(map(int,input("Enter the numbers with spaces").split(" ")))
positive,negative,zero = count_number_types(numbers)
print("Positive numbers: ",positive)
print("Negative numbers: ",negative)
print("zeros: ",zero)
```

```python
# Assignment 34: Prime Number Function
# Write a function named is_prime().

# Requirements
# Accept one integer.
# Return True when the number is prime.
# Return False otherwise.
# Use the function to print all prime numbers from 1 to 100.

def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


for i in range(1, 101):
    if is_prime(i):
        print(i, end=" ")


```

```python
# Assignment 35: Fibonacci Function
# Write a function named generate_fibonacci().

# Requirements
# Accept the number of terms.
# Return the complete Fibonacci series as a list.
# Handle zero and negative term counts.
# Do not use recursion in this assignment.

def generate_fibonacci(number):
    if number <= 0:
        return []

    fib = []
    a, b = 0, 1

    for i in range(number):
        fib.append(a)
        a, b = b, a + b

    return fib


terms = int(input("Enter number of terms: "))

print(generate_fibonacci(terms))
```

```python
# Assignment 67: Filter Prime Numbers
# Accept a list of integers.

# Requirements
# Write a reusable is_prime() function.
# Pass the function to filter().
# Print only prime values.
# Handle values below 2 correctly.

def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

prime_numbers = list(filter(is_prime, numbers))

print("Original List:", numbers)
print("Prime Numbers:", prime_numbers)
```

```python
# Assignment 69: Filter Valid Email Addresses
# Given a list of email strings:

# Requirements
# Use filter().
# A basic valid email should:
# Contain exactly one @.
# Have characters before @.
# Contain a dot after @.
# Return only emails satisfying these conditions.
# Note
# A complete regular expression is not required.

def is_valid_email(email):
    if email.count("@") != 1:
        return False

    username, domain = email.split("@")

    if username == "":
        return False

    if "." not in domain:
        return False

    return True


emails = input("Enter email addresses separated by spaces: ").split()

valid_emails = list(filter(is_valid_email, emails))

print("Valid Emails:", valid_emails)
```

```python
# Assignment 81: Calculate Factorial Using reduce()
# Write a function that calculates factorial using reduce().

# Requirements
# Accept a non-negative integer.
# Generate numbers from 1 to the given number.
# Multiply them using reduce().
# Return 1 when the input is 0.
# Reject negative values.

def factorial(number):
    if number < 0:
        return "Factorial is not defined for negative numbers"

    if number == 0:
        return 1

    numbers = range(1, number + 1)

    return reduce(
        lambda a, b: a * b,
        numbers,
        1
    )

number = int(input("Enter a number: "))

print("Factorial:", factorial(number))
```

```python
# Assignment 90: Closure-Based Discount Calculator
# Write a function named create_discount().

# Requirements
# The outer function should accept a discount percentage.
# The inner function should accept a product price.
# Return the final price after applying the discount.
# Create separate functions for 10%, 20%, and 30% discounts.

def create_discount(discount_percentage):

    def apply_discount(price):
        return price - (price * discount_percentage / 100)

    return apply_discount

discount_10 = create_discount(10)
discount_20 = create_discount(20)
discount_30 = create_discount(30)

price = float(input("Enter product price: "))

print("10% Discount Price:", discount_10(price))
print("20% Discount Price:", discount_20(price))
print("30% Discount Price:", discount_30(price))
```

```python
# Assignment 91: Closure-Based Counter
# Write a function named create_counter().

# Requirements
# Define a variable count = 0 in the outer function.
# Define an inner function that increases the count.
# Use nonlocal to modify the enclosing variable.
# Return the inner function.
# Call the returned function at least five times.

def create_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

my_counter = create_counter()

print(my_counter())
print(my_counter())
print(my_counter())
print(my_counter())
print(my_counter())
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# Requirements
# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry
# Use nonlocal to update the balance.
# Reject withdrawals greater than the available balance.
# Return all three inner functions.

def create_bank_account(balance):

    def deposit(amount):
        nonlocal balance
        balance += amount
        print("Deposited:", amount)

    def withdraw(amount):
        nonlocal balance

        if amount > balance:
            print("Insufficient Balance")
        else:
            balance -= amount
            print("Withdrawn:", amount)

    def check_balance():
        print("Available Balance:", balance)

    return deposit, withdraw, check_balance

deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()

deposit(500)
check_balance()

withdraw(300)
check_balance()

withdraw(2000)
check_balance()
```

```python
# Assignment 93: Function Decorator
# Create a decorator named execution_logger().

# Requirements
# The decorator should print a message before the original function executes.
# It should print another message after execution.
# Apply it to an addition function.
# Preserve the original function's return value.

def execution_logger(func):

    def wrapper(*args, **kwargs):
        print("Starting function...")

        result = func(*args, **kwargs)

        print("Result:", result)
        print("Function completed.")

        return result

    return wrapper


@execution_logger
def add(a, b):
    return a + b


add(10, 20)
```

```python
# Assignment 97: Enclosing Scope Demonstration
# Write an outer function and an inner function.

# Requirements
# Declare a variable named message in the outer function.
# Access it from the inner function.
# Do not pass it as an argument.
# Explain in comments why this is called enclosing scope.

def outer_function():
    message = "Hello from the outer function!"

    def inner_function():
        print(message)

    inner_function()


outer_function()

# Explanation:
# 'message' is defined inside outer_function().
# It is not a local variable of inner_function(),
# but inner_function() can still access it.
# This is called Enclosing Scope because the variable
# belongs to the enclosing (outer) function.
# Python follows LEGB Scope:
# L = Local
# E = Enclosing
# G = Global
# B = Built-in
```

```python
# Assignment 100: Complete Functional Programming Project
# Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.

# Sample Employee Data
# employees = [
#     {
#         "id": 101,
#         "name": "Ravi",
#         "department": "IT",
#         "salary": 60000,
#         "performance_score": 88,
#         "experience": 5
#     },
#     {
#         "id": 102,
#         "name": "Anitha",
#         "department": "HR",
#         "salary": 45000,
#         "performance_score": 72,
#         "experience": 4
#     },
#     {
#         "id": 103,
#         "name": "Kiran",
#         "department": "IT",
#         "salary": 75000,
#         "performance_score": 94,
#         "experience": 7
#     },
#     {
#         "id": 104,
#         "name": "Meena",
#         "department": "Finance",
#         "salary": 55000,
#         "performance_score": 81,
#         "experience": 6
#     },
#     {
#         "id": 105,
#         "name": "Arjun",
#         "department": "IT",
#         "salary": 40000,
#         "performance_score": 65,
#         "experience": 2
#     }
# ]
# Functional Requirements
# Part 1: Display Employees
# Write a function named display_employees() that prints all employee details in a readable format.

# Part 2: Extract Employee Names
# Use map() to extract employee names.

# Part 3: Filter High Performers
# Use filter() to select employees whose performance score is 80 or above.

# Part 4: Filter by Department
# Write a function that accepts a department name and returns matching employees using filter().

# Part 5: Salary Increase
# Use map() to calculate revised salaries based on these conditions:

# Performance score 90 or above: 15% increase
# Performance score 80–89: 10% increase
# Performance score 70–79: 5% increase
# Below 70: No increase
# Do not modify the original employee dictionaries directly.

# Part 6: Total Salary Expense
# Use reduce() to calculate:

# Current total salary
# Revised total salary
# Total additional salary expense
# Part 7: Highest-Paid Employee
# Use reduce() to find the employee with the highest salary.

# Do not use max().

# Part 8: Sort Employees
# Use sorted() with lambda expressions to sort employees by:

# Name
# Salary
# Performance score
# Experience
# Part 9: Closure-Based Bonus Calculator
# Create a function named create_bonus_calculator().

# The outer function should accept a bonus percentage.
# The inner function should accept a salary.
# Return the calculated bonus.
# Create separate bonus calculators for 5%, 10%, and 15%.
# Part 10: Nested Report Generator
# Create an outer function named generate_department_report().

# Inside it, define nested functions to calculate:

# Number of employees
# Total salary
# Average salary
# Average performance score
# Return the complete department report as a dictionary.

# Part 11: LEGB Demonstration
# Create a global variable:

# company_name = "Edukron"
# Inside the report function:

# Access the global company name.
# Create an enclosing variable for the department.
# Create a local variable for the report title.
# Add comments explaining how Python resolves each variable using LEGB.
# Part 12: Menu-Driven Program
# Create a menu containing the following options:

# 1. Display all employees
# 2. Display employee names
# 3. Display high performers
# 4. Search by department
# 5. Calculate revised salaries
# 6. Calculate total salary expense
# 7. Find highest-paid employee
# 8. Sort employees
# 9. Generate department report
# 10. Exit
# Validation Requirements
# Handle invalid menu choices.
# Handle department names that do not exist.
# Handle empty employee lists.
# Validate salary and performance-score values.
# Keep displaying the menu until the user selects Exit.
# Final Submission Requirements
# The student must submit:

# Complete executable Python code.
# Comments explaining every major section.
# At least three sample executions.
# Sample input and output.
# A brief explanation of:
# Functions
# Lambda expressions
# map()
# filter()
# reduce()
# Nested functions
# Closures
# Local scope
# Enclosing scope
# Global scope
# Built-in scope
# A short comparison of normal fu# Python Functions, Map, Filter, Reduce, Nested Functions, Lambda and LEGB Scope


from functools import reduce

# -------------------------------
# Global Variable (LEGB Demo)
# -------------------------------
company_name = "Edukron"

employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2}
]

# --------------------------------------
# Part 1 : Display Employees
# --------------------------------------
def display_employees():
    for emp in employees:
        print(emp)

# --------------------------------------
# Part 2 : Extract Names Using map
# --------------------------------------
def get_employee_names():
    return list(map(lambda emp: emp["name"], employees))

# --------------------------------------
# Part 3 : High Performers
# --------------------------------------
def high_performers():
    return list(filter(lambda emp: emp["performance_score"] >= 80, employees))

# --------------------------------------
# Part 4 : Department Search
# --------------------------------------
def department_filter(department):
    return list(filter(
        lambda emp: emp["department"].lower() == department.lower(),
        employees
    ))

# --------------------------------------
# Part 5 : Revised Salaries
# --------------------------------------
def revised_salaries():

    def calculate(emp):

        increase = 0

        if emp["performance_score"] >= 90:
            increase = 15

        elif emp["performance_score"] >= 80:
            increase = 10

        elif emp["performance_score"] >= 70:
            increase = 5

        revised_salary = emp["salary"] + (
            emp["salary"] * increase / 100
        )

        return {
            "name": emp["name"],
            "old_salary": emp["salary"],
            "new_salary": revised_salary
        }

    return list(map(calculate, employees))

# --------------------------------------
# Part 6 : Salary Expense Using reduce
# --------------------------------------
def salary_expense():

    current_total = reduce(
        lambda total, emp: total + emp["salary"],
        employees,
        0
    )

    revised = revised_salaries()

    revised_total = reduce(
        lambda total, emp: total + emp["new_salary"],
        revised,
        0
    )

    additional = revised_total - current_total

    print("Current Total Salary:", current_total)
    print("Revised Total Salary:", revised_total)
    print("Additional Expense:", additional)

# --------------------------------------
# Part 7 : Highest Paid Employee
# --------------------------------------
def highest_paid():

    return reduce(
        lambda a, b:
        a if a["salary"] > b["salary"] else b,
        employees
    )

# --------------------------------------
# Part 8 : Sorting
# --------------------------------------
def sort_by_name():
    return sorted(employees,
                  key=lambda emp: emp["name"])

def sort_by_salary():
    return sorted(employees,
                  key=lambda emp: emp["salary"])

def sort_by_score():
    return sorted(employees,
                  key=lambda emp:
                  emp["performance_score"])

def sort_by_experience():
    return sorted(employees,
                  key=lambda emp:
                  emp["experience"])

# --------------------------------------
# Part 9 : Closure Bonus Calculator
# --------------------------------------
def create_bonus_calculator(percent):

    def calculate_bonus(salary):
        return salary * percent / 100

    return calculate_bonus

bonus5 = create_bonus_calculator(5)
bonus10 = create_bonus_calculator(10)
bonus15 = create_bonus_calculator(15)

# --------------------------------------
# Part 10 : Nested Department Report
# --------------------------------------
def generate_department_report(dept):

    department_name = dept

    department_data = department_filter(dept)

    report_title = f"{dept} Department Report"

    def employee_count():
        return len(department_data)

    def total_salary():
        return reduce(
            lambda total, emp:
            total + emp["salary"],
            department_data,
            0
        )

    def average_salary():

        if len(department_data) == 0:
            return 0

        return total_salary() / len(department_data)

    def average_score():

        if len(department_data) == 0:
            return 0

        total = reduce(
            lambda acc, emp:
            acc + emp["performance_score"],
            department_data,
            0
        )

        return total / len(department_data)

    return {
        "company": company_name,
        "department": department_name,
        "report_title": report_title,
        "employees": employee_count(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_score": average_score()
    }

# --------------------------------------
# Part 12 : Menu Program
# --------------------------------------
while True:

    print("\nEmployee Performance System")
    print("1. Display Employees")
    print("2. Employee Names")
    print("3. High Performers")
    print("4. Search Department")
    print("5. Revised Salaries")
    print("6. Salary Expense")
    print("7. Highest Paid")
    print("8. Sort By Name")
    print("9. Department Report")
    print("10. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        display_employees()

    elif choice == "2":
        print(get_employee_names())

    elif choice == "3":
        print(high_performers())

    elif choice == "4":
        dept = input("Department: ")
        result = department_filter(dept)

        if result:
            print(result)
        else:
            print("Department Not Found")

    elif choice == "5":
        print(revised_salaries())

    elif choice == "6":
        salary_expense()

    elif choice == "7":
        print(highest_paid())

    elif choice == "8":
        print(sort_by_name())

    elif choice == "9":
        dept = input("Department: ")
        print(generate_department_report(dept))

    elif choice == "10":
        print("Program Closed")
        break

    else:
        print("Invalid Choice")
```
