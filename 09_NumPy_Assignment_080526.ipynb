```python
# Question 40: Low-Margin Transactions
# Management considers a profit margin below 22% to be weak.

# Find all store-category combinations where the profit margin is below 22%.

# Display:

# Store
# Category
# Sales
# Cost
# Profit
# Profit margin
# Use Boolean indexing.

store_cat_profits = sales_data - cost_data

store_cat_profits_margin = np.round(store_cat_profits / sales_data * 100,2)
#print(store_cat_profits_margin)

low_margin = store_cat_profits_margin < 22
#print(low_margin)

rows,cols = np.where(low_margin)

if len(rows)==0:
  print("All the values are above 22")
else:
  for rows,cols in zip(rows,cols):
    print("Store:", stores[rows])
    print("Category:", categories[cols])
    print("Sales:", sales_data[rows, cols])
    print("Cost:", cost_data[rows, cols])
    print("Profit:", store_cat_profits[rows, cols])
    print("Profit Margin:", store_cat_profits_margin[rows, cols], "%")
```
