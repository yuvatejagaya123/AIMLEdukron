```python
#13. Using filter(), map(), and reduce():

# Extract active employees.
# Increase every active employee's salary by 15%.
# Calculate the total updated payroll.

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]

active_emp = list(filter(lambda x : x["active"]== True, employees))
inc_salary_active_emp = list(map(lambda x : x["salary"] + x["salary"]*15/100,active_emp))
updated_payroll_total = float(reduce(lambda x,y : x + y,inc_salary_active_emp,0 ))
updated_payroll_total
```

```python
#14. Using filter() and reduce():

# Extract students who passed.
# Find the topper among the passed students.

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

pass_students = list(filter(lambda x : x["passed"] == True,students))
topper = reduce(lambda x,y : x if x["marks"] > y["marks"] else y,pass_students)
topper
```

```python
#15. Using filter(), map(), and reduce():

# Extract students scoring more than 70.
# Extract only their marks.
# Calculate the average marks.

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

score_student = list(filter(lambda x : x['marks'] > 70, students))
marks = list(map(lambda x : x['marks'], score_student))
avg_marks = reduce(lambda x,y:x+7/len(marks),marks)
avg_marks
```

```python
#16. Using filter(), map(), and reduce():

# Extract products currently in stock.
# Calculate each product's inventory value using price × stock.
# Calculate the total inventory value.

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

in_stock = list(filter(lambda x : x["stock"] > 0,products))
inventory = list(map(lambda x : x["price"] * x["stock"],in_stock))
total_inv = reduce(lambda x,y : x + y, inventory)
total_inv
```

```python
#17. Using filter(), map(), and reduce():

# Extract products belonging to the Electronics category.
# Apply a 20% discount.
# Calculate the total discounted value.

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

ele_products = list(filter(lambda x : x["category"] == 'Electronics',products))
discount = list(map(lambda x : x["price"] - x["price"]*20/100, ele_products))
total_discount = reduce(lambda x,y : x + y,discount)
total_discount
```

```python
#19. Using filter(), map(), and reduce():

# Extract transactions whose amount is greater than 5000.
# Add 18% GST to every selected transaction.
# Calculate the grand total after GST.

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

greater_trans = list(filter(lambda x : x["amount"] > 5000, transactions ))
gst_add = list(map(lambda x : x["amount"] + x["amount"]* 18/100,greater_trans))
total_withgst = reduce(lambda x,y : x + y,gst_add,0 )
total_withgst
```

```python
#20. Using filter(), map(), and reduce():

# Extract active salespeople from the South region.
# Calculate a 5% incentive for each selected salesperson.
# Create a list containing salesperson name, sales amount, and incentive.
# Calculate the total incentive payable.

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

salespeople_south = list(filter(lambda x : x["region"] == "South" and x["active"] == True,sales_records))
incentive = list(map(lambda x : {
    "salesperson": x["salesperson"],
            "sales": x["sales"],
            "incentive": round(x["sales"] * 0.05, 2)},salespeople_south))
print("Salesperson List:")
print(*incentive, sep="\n")
total_incentive_payable = reduce(lambda x,y : x + y["incentive"],incentive,0)
print(f"total incantive payble: {total_incentive_payable}")
```
