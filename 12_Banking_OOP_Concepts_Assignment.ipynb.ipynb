Q34. Compose transfer from transferline objects for banking case 1; use sample reference ACC-1034 and explain the result.

Q72. Implement an observer for low balance for banking case 1; use sample reference ACC-1072 and explain the result.

```python
class SMSALERT:
  def update(self,Account_number,Balance):
    print(f"SMS Alert: {Account_number} has a low balance of {Balance}")

class EMAILAlert:
  def update(self,Account_number,Balance):
    print(f"Email Alert: {Account_number} has a low balance of {Balance}")

class Account:
  def __init__(self,Account_number,Balance,Minimum_balance):
    self.Account_number = Account_number
    self.Balance = Balance
    self.Minimum_balance = Minimum_balance
    self.observers = []

  def add_observers(self,observer):
    self.observers.append(observer)

  def remove_observers(self,observer):
    if observer in self.observers:
      self.observers.remove(observer)

  def Notify(self):
    for observer in self.observers:
      observer.update(self.Account_number,self.Balance)

  def withdraw(self,amount):
    if amount <= 0:
      print(" Amount should be greater than zero")
    elif amount > self.Balance:
      print("Insufficent Balance")
    else:
      self.Balance -= amount

      print(f"{amount} Withdrawn Successfully")
      print(f"Current Balance:{self.Balance}")

      if self.Balance < self.Minimum_balance:
        self.Notify()

account = Account("ACC-1072",100000,5000)

sms_alert = SMSALERT()
Email_alert = EMAILAlert()

account.add_observers(sms_alert)
account.add_observers(Email_alert)

account.withdraw(96000)

### The Observer Pattern creates a one-to-many relationship between objects. When the subject's state changes, all registered observers are notified automatically.
### Here, the Account is the subject, and SMSAlert and EmailAlert are observers that receive notifications when the account balance falls below the minimum balance.
```

Q75. Implement a unit-of-work transaction for banking case 1; use sample reference ACC-1075 and explain the result.

```python
class Account:
    def __init__(self, account_number, account_name, balance):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance

    def __str__(self):
        return (f"{self.account_number} | {self.account_name} | Balance: ₹{self.balance}")


class AccountRepository:
    def __init__(self):
        self.accounts = {}

    def add(self, account):
        self.accounts[account.account_number] = account

    def get_by_id(self, account_number):
        return self.accounts.get(account_number)

class UnitOfWork:
    def __init__(self, repository):
        self.repository = repository
        self.original_balances = {}
        self.is_committed = False

    def __enter__(self):
        # Save original balances before making changes
        for account_number, account in self.repository.accounts.items():
            self.original_balances[account_number] = account.balance

        print("Transaction started.")
        return self

    def commit(self):
        self.is_committed = True
        print("Transaction committed successfully.")

    def rollback(self):
        # Restore every account's original balance
        for account_number, original_balance in self.original_balances.items():
            account = self.repository.get_by_id(account_number)

            if account is not None:
                account.balance = original_balance

        print("Transaction rolled back.")

    def __exit__(self, exception_type, exception_value, traceback):
        if exception_type is not None or not self.is_committed:
            self.rollback()

        # False means Python will not hide an exception
        return False


class TransferService:
    def transfer(self, from_account, to_account, amount):
        if amount <= 0:
            raise ValueError("Transfer amount must be greater than zero.")

        if from_account.balance < amount:
            raise ValueError("Insufficient balance in sender account.")

        from_account.balance -= amount
        to_account.balance += amount

        print(f"₹{amount} transferred from {from_account.account_number} to {to_account.account_number}.")


# Create repository
repository = AccountRepository()

# Create accounts
sender = Account("ACC-1075","Yuvateja",10000)

receiver = Account("ACC-2001","Rahul",5000)

# Store accounts in repository
repository.add(sender)
repository.add(receiver)

print("Balances before transfer:")
print(sender)
print(receiver)

transfer_service = TransferService()

# Unit-of-work transaction
try:
    with UnitOfWork(repository) as unit_of_work:
        transfer_service.transfer(sender,receiver,3000)
        unit_of_work.commit()

except ValueError as error:
    print("Error:", error)

print("\nBalances after transfer:")
print(sender)
print(receiver)

### The Unit of Work pattern groups multiple data changes into a single transaction. It commits all changes when every operation succeeds and rolls them back when any operation fails.
### In this example, debiting ACC-1075 and crediting the receiver are treated as one atomic transaction.
```

Q76. Separate domain and infrastructure layers for banking case 1; use sample reference ACC-1076 and explain the result.

```python
# ==================================================
# DOMAIN LAYER
# Contains banking business rules
# ==================================================

class Account:
    def __init__(self, account_number, account_name, balance):
        if balance < 0:
            raise ValueError("Opening balance cannot be negative.")

        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be greater than zero.")

        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be greater than zero.")

        if amount > self.balance:
            raise ValueError("Insufficient balance.")

        self.balance -= amount

    def __str__(self):
        return (f"{self.account_number} | {self.account_name} | Balance: ₹{self.balance}")


class AccountRepository:
    """Domain-level repository abstraction."""

    def save(self, account):
        raise NotImplementedError("Subclass must implement save().")

    def get_by_id(self, account_number):
        raise NotImplementedError("Subclass must implement get_by_id().")


# ==================================================
# INFRASTRUCTURE LAYER
# Contains technical storage implementation
# ==================================================

class InMemoryAccountRepository(AccountRepository):
    def __init__(self):
        self._accounts = {}

    def save(self, account):
        self._accounts[account.account_number] = account

    def get_by_id(self, account_number):
        return self._accounts.get(account_number)


# ==================================================
# APPLICATION / EXECUTION LAYER
# Connects domain and infrastructure
# ==================================================

repository = InMemoryAccountRepository()

account = Account(account_number="ACC-1076",account_name="Yuvateja",balance=10000)

account.deposit(3000)
account.withdraw(2000)

repository.save(account)

saved_account = repository.get_by_id("ACC-1076")

if saved_account is not None:
    print(saved_account)
else:
    print("Account not found.")

### Layer separation keeps banking business rules independent from technical storage details. The domain layer contains the Account entity and its business validations, while the infrastructure layer implements data storage through InMemoryAccountRepository.
### This allows the storage technology to change without modifying the core banking logic.
```

Q85. Implement a unit-of-work transaction for banking case 2; use sample reference ACC-1085 and explain the result.

```python
class Account:
    def __init__(self, account_number, account_name, balance):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance

    def __str__(self):
        return (f"{self.account_number} | {self.account_name} | Balance: ₹{self.balance}")


class AccountRepository:
    def __init__(self):
        self.accounts = {}

    def add(self, account):
        self.accounts[account.account_number] = account

    def get_by_id(self, account_number):
        return self.accounts.get(account_number)

class UnitOfWork:
    def __init__(self, repository):
        self.repository = repository
        self.original_balances = {}
        self.is_committed = False

    def __enter__(self):
        # Save original balances before making changes
        for account_number, account in self.repository.accounts.items():
            self.original_balances[account_number] = account.balance

        print("Transaction started.")
        return self

    def commit(self):
        self.is_committed = True
        print("Transaction committed successfully.")

    def rollback(self):
        # Restore every account's original balance
        for account_number, original_balance in self.original_balances.items():
            account = self.repository.get_by_id(account_number)

            if account is not None:
                account.balance = original_balance

        print("Transaction rolled back.")

    def __exit__(self, exception_type, exception_value, traceback):
        if exception_type is not None or not self.is_committed:
            self.rollback()

        # False means Python will not hide an exception
        return False


class TransferService:
    def transfer(self, from_account, to_account, amount):
        if amount <= 0:
            raise ValueError("Transfer amount must be greater than zero.")

        if from_account.balance < amount:
            raise ValueError("Insufficient balance in sender account.")

        from_account.balance -= amount
        to_account.balance += amount

        print(f"₹{amount} transferred from {from_account.account_number} to {to_account.account_number}.")


# Create repository
repository = AccountRepository()

# Create accounts
sender = Account("ACC-1085","Yuvateja",10000)

receiver = Account("ACC-2001","Rahul",5000)

# Store accounts in repository
repository.add(sender)
repository.add(receiver)

print("Balances before transfer:")
print(sender)
print(receiver)

transfer_service = TransferService()

# Unit-of-work transaction
try:
    with UnitOfWork(repository) as unit_of_work:
        transfer_service.transfer(sender,receiver,3000)
        unit_of_work.commit()

except ValueError as error:
    print("Error:", error)

print("\nBalances after transfer:")
print(sender)
print(receiver)

### The Unit of Work pattern groups multiple data changes into a single transaction. It commits all changes when every operation succeeds and rolls them back when any operation fails.
### In this example, debiting ACC-1085 and crediting the receiver are treated as one atomic transaction.
```

Q97. Design a plugin-ready payment architecture for banking case 3; use sample reference ACC-1097 and explain the result.

```python
class UPIPlugin:
    def pay(self, account_reference, amount):
        return (f"Payment successful: ₹{amount} paid through UPI for {account_reference}")


class CreditCardPlugin:
    def pay(self, account_reference, amount):
        return (f"Payment successful: ₹{amount} paid through Credit Card for {account_reference}")


class PaymentProcessor:
    def __init__(self):
        self.plugins = {}

    def register_plugin(self, payment_name, payment_plugin):
        self.plugins[payment_name.upper()] = payment_plugin

    def process_payment(self,payment_name,account_reference,amount):
        if amount <= 0:
            raise ValueError("Payment amount must be greater than zero.")

        payment_name = payment_name.upper()
        payment_plugin = self.plugins.get(payment_name)

        if payment_plugin is None:
            raise ValueError(f"Payment plugin '{payment_name}' is not registered.")

        return payment_plugin.pay(account_reference,amount)


# Create the payment processor
processor = PaymentProcessor()

# Register available payment plugins
processor.register_plugin("UPI",UPIPlugin())

processor.register_plugin("CREDIT_CARD",CreditCardPlugin())

# Process a payment
result = processor.process_payment(payment_name="UPI",account_reference="ACC-1097",amount=5000)

print(result)

### A plugin-ready payment architecture defines a common operation for all payment plugins and maintains a registry of available implementations. The processor selects the required plugin at runtime.
### New payment methods can be introduced by creating and registering additional plugin classes without changing the payment processor.
```

Q99. Use cooperative inheritance with super for banking case 3; use sample reference ACC-1099 and explain the result.

```python
class Base:
    def __init__(self, **kwargs):
        super().__init__(**kwargs)


class AccountInfo(Base):
    def __init__(self,account_number,account_name,**kwargs):
        self.account_number = account_number
        self.account_name = account_name

        super().__init__(**kwargs)


class BalanceMixin(Base):
    def __init__(self,balance,**kwargs):
        self.balance = balance

        super().__init__(**kwargs)


class BankingAccount(AccountInfo,BalanceMixin):
    def __init__(self,account_number,account_name,balance):
        super().__init__(account_number=account_number,account_name=account_name,balance=balance)

    def display(self):
        print(f"{self.account_number} | {self.account_name} | ₹{self.balance}")


account = BankingAccount("ACC-1099","Yuvateja",15000)

account.display()

### Cooperative inheritance uses super() so that all classes in a multiple inheritance hierarchy participate in initialization according to Python's Method Resolution Order (MRO).
### This avoids duplicate constructor calls and makes the inheritance chain easier to maintain.
```
