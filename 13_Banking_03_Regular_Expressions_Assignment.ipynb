Q31. Parse a receipt line into named groups for banking case 1; use sample reference ACC-1031 and explain the result.

```python
import re

Receipt_line = "ACC-1031 | DEPOSIT | 6000.98 | 09-09-2026"

pattern = r"^(?P<account>ACC-\d{4})\s*\|\s*(?P<transactiontype>[A-Z]+)\s*\|\s*(?P<amount>\d+(?:\.\d{1,2})?)\s*\|\s*(?P<date>\d{2}-\d{2}-\d{4})$"

# \d+ → one or more digits
# (?:...) → non-capturing group
# \. → decimal point
# \d{1,2} → one or two decimal digits
# ? → makes the decimal portion optional

# pattern = (
#     r"^(?P<account>ACC-\d{4})\s*\|\s*"
#     r"(?P<transaction>[A-Z]+)\s*\|\s*"
#     r"(?P<amount>\d+(?:\.\d{1,2})?)\s*\|\s*"
#     r"(?P<date>\d{2}-\d{2}-\d{4})$"
# )

match = re.fullmatch(pattern,Receipt_line)

if match:
  print("Receipt Parsed Successfully")
  print("Account     :",match.group("account"))
  print("Transaction :", match.group("transactiontype"))
  print("Amount      :", match.group("amount"))
  print("Date        :", match.group("date"))

  print("\nNamed groups:", match.groupdict())
else:
  print("Invalid receipt format")


```

Q44. Compile a reusable account pattern for banking case 1; use sample reference ACC-1044 and explain the result.

```python
import re

Acc_num = "ACC-1044"

accounts = [
    "ACC-1001",
    "ACC-2050",
    "ACC-9999",
    "ACC1001",
    "acc-1234",
    "ACC-12"
]

account_pattern = re.compile(r"^ACC-\d{4}$")

print("Reference:", Acc_num)
print("-" * 30)

for acc in accounts:
    if account_pattern.fullmatch(acc):
        print(acc, "-> Valid Account")
    else:
        print(acc, "-> Invalid Account")
#### Pattern is compiled once and reused multiple times, which is cleaner and faster when validating many account references.

### Use re.compile() when the same regex will be used repeatedly inside loops, validations, log parsing, or data-processing pipelines.
```

Q80. Create a regex-based data quality report for banking case 1; use sample reference ACC-1080 and explain the result.

```python
import re

Acc_num = "ACC-1080"

records = [
    {
        "account": "ACC-1001",
        "email": "john@gmail.com",
        "phone": "9876543210"
    },
    {
        "account": "ACC1002",
        "email": "invalid@email",
        "phone": "98765"
    },
    {
        "account": "ACC-1003",
        "email": "support@banking.co.in",
        "phone": "9123456789"
    }
]

account_pattern = r"^ACC-\d{4}$"
email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
phone_pattern = r"^[6-9]\d{9}$"

valid_records = 0

for record in records:

    account_valid = bool(re.fullmatch(account_pattern, record["account"]))

    email_valid = bool(re.fullmatch(email_pattern, record["email"]))

    phone_valid = bool(re.fullmatch(phone_pattern, record["phone"]))

    status = (account_valid and email_valid and phone_valid)

    if status:
        valid_records += 1

    print("\nRecord:", record)
    print("Account :", account_valid)
    print("Email   :", email_valid)
    print("Phone   :", phone_valid)
    print("Status  :", "PASS" if status else "FAIL")

print("\n----- DATA QUALITY REPORT -----")
print("Total Records :", len(records))
print("Valid Records :", valid_records)
print("Invalid Records :", len(records) - valid_records)
```

Q97. Build a verbose-mode pattern for account holder ids for banking case 3; use sample reference ACC-1097 and explain the result.

Q98. Avoid catastrophic backtracking in a interest offer pattern for banking case 3; use sample reference ACC-1098 and explain the result.

```python
import re

Acc_num = "ACC-1098"

offer = "INTEREST-OFFER-15"

pattern = r"^[A-Z]+(?:-[A-Z]+)*-\d+$"

result = re.fullmatch(pattern, offer)

if result:
    print(f"{Acc_num} : Valid Interest Offer")
else:
    print(f"{Acc_num} : Invalid Interest Offer")

 ### Catastrophic backtracking occurs when a regex contains nested or overlapping quantifiers, causing excessive retry attempts during matching.
 ### To avoid it, use specific character classes, anchors, and non-ambiguous repetitions instead of patterns like (.+)+ or (.*)*.
  ### For ACC-1078, ^[A-Z]+(?:-[A-Z]+)*-\d+$ is a safe and efficient interest-offer validation pattern.
```

Q100. Create a regex-based data quality report for banking case 3; use sample reference ACC-1100 and explain the result.

```python
import re

Acc_num = "ACC-1090"

records = [
    {
        "account": "ACC-1001",
        "email": "john@gmail.com",
        "phone": "9876543210"
    },
    {
        "account": "ACC1002",
        "email": "invalid@email",
        "phone": "98765"
    },
    {
        "account": "ACC-1003",
        "email": "support@banking.co.in",
        "phone": "9123456789"
    }
]

account_pattern = r"^ACC-\d{4}$"
email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
phone_pattern = r"^[6-9]\d{9}$"

valid_records = 0

for record in records:

    account_valid = bool(re.fullmatch(account_pattern, record["account"]))

    email_valid = bool(re.fullmatch(email_pattern, record["email"]))

    phone_valid = bool(re.fullmatch(phone_pattern, record["phone"]))

    status = (account_valid and email_valid and phone_valid)

    if status:
        valid_records += 1

    print("\nRecord:", record)
    print("Account :", account_valid)
    print("Email   :", email_valid)
    print("Phone   :", phone_valid)
    print("Status  :", "PASS" if status else "FAIL")

print("\n----- DATA QUALITY REPORT -----")
print("Total Records :", len(records))
print("Valid Records :", valid_records)
print("Invalid Records :", len(records) - valid_records)
```
